```python
import pandas as pd

df = pd.read_csv("C:/Users/HIII/banking-fraud-detection/data/creditcard.csv")

df.head()
```

```python
df["Class"].value_counts()
```

```python
fraud_percentage = (df["Class"].sum() / len(df)) * 100

print("Fraud Percentage:", fraud_percentage)
```

```python
df.isnull().sum()
```

```python
import matplotlib.pyplot as plt

df["Class"].value_counts().plot(kind="bar")

plt.title("Fraud vs Normal Transactions")
plt.xlabel("Class")
plt.ylabel("Count")

plt.show()
```

```python
df["Amount"].describe()
```

```python
fraud = df[df["Class"] == 1]
normal = df[df["Class"] == 0]

print("Fraud Transaction Amounts")
print(fraud["Amount"].describe())

print("\nNormal Transaction Amounts")
print(normal["Amount"].describe())
```

```python
plt.figure(figsize=(10,5))

plt.hist(normal["Amount"], bins=50, alpha=0.5, label="Normal")
plt.hist(fraud["Amount"], bins=50, alpha=0.5, label="Fraud")

plt.title("Transaction Amount Distribution")
plt.xlabel("Amount")
plt.ylabel("Frequency")
plt.legend()

plt.show()
```

```python
import seaborn as sns

plt.figure(figsize=(12,10))

sns.heatmap(df.corr(), cmap="coolwarm")

plt.title("Correlation Heatmap")

plt.show()
```

```python
plt.figure(figsize=(10,5))

plt.hist(df["Time"], bins=50)

plt.title("Transaction Time Distribution")
plt.xlabel("Time")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(x=df["Amount"])

plt.title("Transaction Amount Boxplot")

plt.show()
```
